Construção da métrica que quantifica a demanda por trabalhadores do TI

1 - Irei carregar as colunas relevantes ('subsetor_ibge_valor', 'ano', 'faixa_salarial', 'valor_remuneracao_media', 'descricao')

2 - Vamos dividir em 2 dataframes, o 1º que retorne a distribuição de funcionário por subsetor ao longo do ano e 2º que retorne a distribuição de funções ao longo do ano.

3 - Vamos tirar a média salarial de cada distribuição (std e amplitude).

4 - Iremos tirar o desvio padrão das médias salariais ao longo dos anos.

Os setores com maior crescimento do desvio padrão vão representar os setores que mais tem apresentado uma maior demanda ao longo dos anos.

```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
from google.colab import drive
drive.mount('/content/drive')
```

```python


drive_file_path = '/content/drive/MyDrive/RESIDÊNCIA /Imersão /data/df_principal2.csv'

try:
    df_drive = pd.read_csv(drive_file_path)
    df = df_drive.copy()
    print("Dataset carregado com sucesso do Google Drive!")
    display(df)
except FileNotFoundError:
    print(f"Erro: O arquivo não foi encontrado em {drive_file_path}. Verifique o caminho.")
except Exception as e:
    print(f"Ocorreu um erro ao carregar o dataset: {e}")
```

```python
demanda_subsetor = ['subsetor_ibge_valor', 'ano', 'faixa_salarial', 'valor_remuneracao_media']
df_demanda_subsetor = df[demanda_subsetor].copy()

display(df_demanda_subsetor)
```

```python
df_demanda_subsetor = df_demanda_subsetor[~df_demanda_subsetor['valor_remuneracao_media'].isin([0])]
```

```python
zero_values = df_demanda_subsetor[df_demanda_subsetor['valor_remuneracao_media'] == 0]['valor_remuneracao_media'].value_counts()
print(zero_values)
```

```python
df_medianas_anuais = df_demanda_subsetor.groupby(['subsetor_ibge_valor', 'ano'])['valor_remuneracao_media'].median().reset_index()
display(df_medianas_anuais)
```

```python
# --- 3. Preparar dados para o CAGR (Início e Fim) ---
# Precisamos do valor inicial (2023) e final (2024) para cada subsetor
ano_inicial = 2022
ano_final = 2024
n_anos = ano_final - ano_inicial # (2024 - 2023 = 1)

# Filtra apenas os anos que nos interessam
df_medianas_inicio = df_medianas_anuais[df_medianas_anuais['ano'] == ano_inicial]
df_medianas_fim = df_medianas_anuais[df_medianas_anuais['ano'] == ano_final]
```

```python
df_cagr = pd.merge(df_medianas_inicio, df_medianas_fim, on='subsetor_ibge_valor', suffixes=('_inicio', '_fim'))

df_cagr = df_cagr.rename(columns={
    'valor_remuneracao_media_inicio': f'mediana_{ano_inicial}',
    'valor_remuneracao_media_fim': f'mediana_{ano_final}'
})

df_cagr = df_cagr.drop(columns=['ano_inicio', 'ano_fim'])
```

```python
# --- 4. Calcular o CAGR (Compound Annual Growth Rate) ---
# Fórmula: ((Valor Final / Valor Inicial) ^ (1 / N_Anos)) - 1

# Garante que o valor inicial não é zero para evitar divisão por zero
df_cagr = df_cagr[df_cagr[f'mediana_{ano_inicial}'] > 0]

df_cagr['cagr_mediana_salarial'] = (
    (df_cagr[f'mediana_{ano_final}'] / df_cagr[f'mediana_{ano_inicial}']) ** (1 / n_anos)
) - 1

# --- 5. Exibir Resultado Final ---
# Ordena do maior CAGR (maior demanda) para o menor
df_resultado_cagr = df_cagr.sort_values(by='cagr_mediana_salarial', ascending=False)

print("\n--- (Resultado) Crescimento Percentual por Subsetor (2022-2024) ---")
display(df_resultado_cagr)

print("\n(Para 1 ano, CAGR = Crescimento Percentual Simples)")
```

```python
demanda_descricao = ['descricao', 'ano', 'faixa_salarial', 'valor_remuneracao_media']
df_demanda_descricao = df[demanda_descricao].copy()

display(df_demanda_descricao)
```

```python
df_demanda_descricao = df_demanda_descricao.dropna(subset=['descricao']).copy()
df_demanda_descricao = df_demanda_descricao[~df_demanda_descricao['valor_remuneracao_media'].isin([0])]
```

```python
zero_values = df_demanda_descricao[df_demanda_descricao['valor_remuneracao_media'] == 0]['valor_remuneracao_media'].value_counts()
print(zero_values)
```

```python
df_medianas_anuais = df_demanda_descricao.groupby(['descricao', 'ano'])['valor_remuneracao_media'].median().reset_index()
display(df_medianas_anuais)
```

```python
# --- 3. Preparar dados para o CAGR (Início e Fim) ---
# Precisamos do valor inicial (2023) e final (2024) para cada subsetor
ano_inicial = 2022
ano_final = 2024
n_anos = ano_final - ano_inicial # (2024 - 2023 = 1)

# Filtra apenas os anos que nos interessam
df_medianas_inicio = df_medianas_anuais[df_medianas_anuais['ano'] == ano_inicial]
df_medianas_fim = df_medianas_anuais[df_medianas_anuais['ano'] == ano_final]
```

```python
df_cagr_descricao = pd.merge(df_medianas_inicio, df_medianas_fim, on='descricao', suffixes=('_inicio', '_fim'))

df_cagr_descricao = df_cagr_descricao.rename(columns={
    'valor_remuneracao_media_inicio': f'mediana_{ano_inicial}',
    'valor_remuneracao_media_fim': f'mediana_{ano_final}'
})

df_cagr_descricao = df_cagr_descricao.drop(columns=['ano_inicio', 'ano_fim'])
```

```python
# --- 4. Calcular o CAGR (Compound Annual Growth Rate) ---
# Fórmula: ((Valor Final / Valor Inicial) ^ (1 / N_Anos)) - 1

# Garante que o valor inicial não é zero para evitar divisão por zero
df_cagr_descricao = df_cagr_descricao[df_cagr_descricao[f'mediana_{ano_inicial}'] > 0]

df_cagr_descricao['cagr_mediana_salarial'] = (
    (df_cagr_descricao[f'mediana_{ano_final}'] / df_cagr_descricao[f'mediana_{ano_inicial}']) ** (1 / n_anos)
) - 1

# --- 5. Exibir Resultado Final ---
# Ordena do maior CAGR (maior demanda) para o menor
df_resultado_cagr_descricao = df_cagr_descricao.sort_values(by='cagr_mediana_salarial', ascending=False)

print("\n--- (Resultado) Crescimento Percentual por Subsetor (2022-2024) ---")
display(df_resultado_cagr_descricao)

print("\n(Para 1 ano, CAGR = Crescimento Percentual Simples)")
```

```python
df_tempo_trabalho = pd.read_csv('/content/drive/MyDrive/RESIDÊNCIA /Imersão /data/df_tempo_trabalho.csv')

print("--- df tempocarregado: ---")
display(df_tempo_trabalho.head())
```

```python
df_analise_tempo = df_tempo_trabalho[df_tempo_trabalho['ano'] >= 2022].copy()

# CRIANDO UMA FLAG PARA SUBSTITUIR O NAN POR 0
colunas_evento = ['mes_admissao', 'mes_desligamento']
df_analise_tempo[colunas_evento] = df_analise_tempo[colunas_evento].fillna(0)

df_analise_tempo['admissao_flag'] = (df_analise_tempo['mes_admissao'] > 0).astype(int)


df_analise_tempo['desligamento_flag'] = (df_analise_tempo['mes_desligamento'] > 0).astype(int)

print("sucesso")
```

```python
# Agrupa por ano, setor e função (descrição)
# E então somamos as flags (1s e 0s) para contar os eventos
df_demanda_tempo = df_analise_tempo.groupby(
    ['ano', 'subsetor_ibge_valor', 'descricao']
).agg(
    total_Admissoes=('admissao_flag', 'sum'),
    total_Desligamentos=('desligamento_flag', 'sum'),
    total_Ativos_Fim_Ano=('vinculo_ativo_3112', 'sum')
).reset_index()
```

```python

# Esta é a métrica de "demanda líquida"
df_demanda_tempo['saldo_vagas'] = df_demanda_tempo['total_Admissoes'] - df_demanda_tempo['total_Desligamentos']

# --- 5. Análise dos Resultados ---

# Ordena para ver os maiores saldos (maior demanda) por ano
df_demanda_tempo_ordenada = df_demanda_tempo.sort_values(
    by=['ano', 'saldo_vagas'],
    ascending=[False, False] # Ano mais recente, maior saldo
)

# Exibe os principais resultados
print("\n--- Top 10 Setores/Funções por saldo de Vagas ---")
display(df_demanda_tempo_ordenada.head(10))
```
