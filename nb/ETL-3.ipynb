```python
df_principal_original = pd.read_csv('C:/Users/Matheus Rosetti/Documents/RESIDENCIA/Imersao/data/df_principal2.csv')

df_principal = df_principal_original.copy()

display(df_principal)
```

```python
df_descricao_por_subsetor_gpy = df_principal.groupby(['ano', 'subsetor_ibge_valor', 'descricao'], dropna=False).agg(
    cont_freq_descricao=('descricao', 'size'),
    media_valor_remuneracao=('valor_remuneracao_media', 'mean')
).reset_index()

display(df_descricao_por_subsetor_gpy)
```

```python
col_agruped_subsetor = 'descricao'
valor_freq_coluna = 'media_valor_remuneracao'

df_descricao_por_subsetor_gpy['std'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('std')
df_descricao_por_subsetor_gpy['min'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('min')
df_descricao_por_subsetor_gpy['max'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('max')
df_descricao_por_subsetor_gpy['q1'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.25))
df_descricao_por_subsetor_gpy['mediana'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.50))
df_descricao_por_subsetor_gpy['q3'] = df_descricao_por_subsetor_gpy.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.75))

display(df_descricao_por_subsetor_gpy)
```

```python
df_descricao_por_subsetor_gpy['ano'] = df_descricao_por_subsetor_gpy['ano'].ffill()
df_descricao_por_subsetor_gpy['subsetor_ibge_valor'] = df_descricao_por_subsetor_gpy['subsetor_ibge_valor'].ffill()
df_descricao_por_subsetor_gpy['descricao'] = df_descricao_por_subsetor_gpy['descricao'].ffill()

df_descricao_por_subsetor_gpy.to_excel("df_descricao_por_subsetor.xlsx", merge_cells=False)
```
