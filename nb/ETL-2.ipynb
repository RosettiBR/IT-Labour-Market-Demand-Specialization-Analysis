```python
# NOVO DATAFRAME COM VALORES AGRUPADOS POR SUBSETOR IBGE
# agrupado primeiro por 'ano' e depois por 'subsetor_ibge'

df_subsetor_describe = df.groupby(['ano', 'tamanho_estabelecimento_valor', 'natureza_juridica_valor', 'subsetor_ibge_valor'], dropna=False).agg(
    cont_freq_subsetor=('subsetor_ibge_valor', 'size'),
    media_valor_remuneracao=('valor_remuneracao_media', 'mean'),
)

print(df_subsetor_describe)
```

```python
# Criando colunas com describe() com base no agrupamento por 'subsetor_ibge_valor' (funcionários)
col_agruped_subsetor = 'subsetor_ibge_valor'
valor_freq_coluna = 'cont_freq_subsetor'

df_subsetor_describe['media'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('mean')
df_subsetor_describe['std'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('std')
df_subsetor_describe['min'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('min')
df_subsetor_describe['max'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform('max')
df_subsetor_describe['q1'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.25))
df_subsetor_describe['mediana'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.50))
df_subsetor_describe['q3'] = df_subsetor_describe.groupby(col_agruped_subsetor)[valor_freq_coluna].transform(lambda x: x.quantile(0.75))

print(df_subsetor_describe)
```
